# Problema 2

```
Clear["Global`*"];
WolframColor = RGBColor[0.941, 0.502, 0.0];
```

```
a1[t_] := 2.2*10^-29*Exp[10^35*t]
a2[t_] := 6.4*10^10*Sqrt[t]
a3[t_] := 7700*t^(4/3)

a[t_] := Piecewise[{
    {a1[t],10^-35 <= t <= 10^-33},
    {a2[t], 10^-33 < t < 9.3*10^12},
    {a3[t], 9.3*10^12 <= t = 4.2*10^17}
    }]
```

## 1

```
Plot[a[t],{t, 1, 10*60},
    PlotLabel -> Style["Inicios de la Era de Radiación", 18, Bold],
    PlotTheme->"Scientific",
    
    FrameTicksStyle -> Directive[14],
    FrameLabel -> {
        Style["t [s]", 18, Bold],   
        Style["a(t)", 18, Bold]},    
        
    ImageSize->600
    ]
```

## 2

```
Plot[a[t],{t, 12*^9, 13*^9},
    PlotLabel -> Style["Inicios de la Era de la Materia", 18, Bold],
    PlotTheme->"Scientific",
    
    FrameTicksStyle -> Directive[14],
    FrameLabel -> {
        Style["t [s]", 18, Bold],   
        Style["a(t)", 18, Bold]},    
        
    ImageSize->600
    ]
```

# Problema 3

```
Clear["Global`*"]
WolframColor = RGBColor[0.941, 0.502, 0.0]
```

## 1

```
h0 = 2;
v = 20;

gTierra = 9.80665;
gMarte = 3.721;
gLuna = 1.625;

H[h0_,v_,g_,t_] := h0+v*t-0.5*g*t^2
```

```
(* Parte de Cálculo *)
VerticeX = Solve[D[H[h0,v,gTierra,t],t]==0,t];

(* Puntos del Vértice *)
X = VerticeX[[1,1,2]] (* Componente x *)
Y = H[h0,v,gTierra,X] (* Componente y *)

(* --------------------------- *)
(* El eje de simetría es x = X *)
(* --------------------------- *)

(* t final --debe ser el positivo-- *)
tFinal = Reduce[H[h0,v,gTierra,t]==0 && t>=0, t][[2]]
```

```
gList   = {gTierra, gMarte, gLuna};
labels  = {"Tierra", "Marte", "Luna"};

Plot[
  Evaluate@Table[
    H[h0, v, g, t],
    {g, gList}
  ],
  {t, 0, tFinal},

  PlotTheme -> "Scientific",

  PlotStyle -> {
    {WolframColor, AbsoluteThickness[2.5], Dashing[{}]},       (* sólida *)
    {WolframColor, AbsoluteThickness[2.5], Dashing[{0.06,0.02}]},   (* punteada *)
    {WolframColor, AbsoluteThickness[2.5], Dashing[{0.02}]} (* raya–punto *)
  },

  PlotLegends -> Placed[
    LineLegend[
      labels,
      LegendMarkers -> None,
      LegendLayout -> "Column"
    ],
    Right
  ],

  Frame -> True,
  FrameLabel -> {
    Style["t [s]", 18, Bold],
    Style["h(t) [m]", 18, Bold]
  },
  FrameTicksStyle -> Directive[16],
  ImageSize -> 600
]
```
