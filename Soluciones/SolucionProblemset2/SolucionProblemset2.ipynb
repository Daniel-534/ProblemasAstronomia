## Problema 2

Debemos despejar $T$

$$T = \sqrt[4]{\frac{F}{5.7\times 10^{-8}}}$$

```
Clear["Global`*"]

c = 5.7*^-8;
F = 1357;


(*Temperatura en Kelvin*)

Style[
  T = (F/c)^(1/4) // TraditionalForm,
  FontSize -> 20
]
```

## Problema 3

```
Clear["Global`*"]

c = 5.7*^-8;

A = 10*100;

(* En MegaWatts *)
F = (c * 2000^4 * A)*10^-6;

Style[F//TraditionalForm,FontSize->20 ]
```

## Problema 4
