## Problema 3

```
Clear["Global`*"]

c = 5.7*^-8;

A = 10*100;

(* En MegaWatts *)
F = (c * 2000^4 * A)*10^-6;

Style[F//TraditionalForm,FontSize->20 ]
```

## Problema 4
