```
(* Versión 1*)
ClearAll["Global`*"];

Bx[x_, y_, z_] := (3*x*z)/(x^2 + y^2 + z^2)^(5/2);
By[x_, y_, z_] := (3*y*z)/(x^2 + y^2 + z^2)^(5/2);
Bz[x_, y_, z_] := (3*z^2 - (x^2 + y^2 + z^2))/(x^2 + y^2 + z^2)^(5/2);

rMin = 0.15;

VectorPlot3D[{Bx[x, y, z], By[x, y, z], Bz[x, y, z]},
  {x, -3, 3}, {y, -3, 3}, {z, -3, 3},
  RegionFunction -> Function[{x, y, z}, x^2 + y^2 + z^2 > rMin^2],
  VectorPoints -> 20,
  VectorScale -> {0.025, Scaled[0.4]},
  VectorColorFunction -> (Blend[{Cyan, Purple, Orange}, #7] &),
  VectorStyle -> Directive[Opacity[0.8], Arrowheads[0.01]],
  AxesLabel -> {"x", "y", "z"},
  BoxRatios -> {1, 1, 1},
  ViewPoint -> {2.4, 1.8, 1.2},
  ImageSize -> 700,
  PlotLabel -> Style["Campo Magnético Terrestre", 14, Italic]]
```

```
(*Versión 2*)
ClearAll["Global`*"];


Bx[x_, y_, z_] := If[x^2 + y^2 + z^2 == 0, 0, (3*x*z)/(x^2 + y^2 + z^2)^(5/2)];
By[x_, y_, z_] := If[x^2 + y^2 + z^2 == 0, 0, (3*y*z)/(x^2 + y^2 + z^2)^(5/2)];
Bz[x_, y_, z_] := If[x^2 + y^2 + z^2 == 0, 0, (3*z^2 - (x^2 + y^2 + z^2))/(x^2 + y^2 + z^2)^(5/2)];

r[x_, y_, z_] := Sqrt[x^2 + y^2 + z^2];

Bx2[x_, y_, z_] := With[{r2 = x^2 + y^2 + z^2}, 
  If[r2 < 10^-6, 0, (3*x*z)/r2^(5/2)]];
By2[x_, y_, z_] := With[{r2 = x^2 + y^2 + z^2}, 
  If[r2 < 10^-6, 0, (3*y*z)/r2^(5/2)]];
Bz2[x_, y_, z_] := With[{r2 = x^2 + y^2 + z^2}, 
  If[r2 < 10^-6, 0, (3*z^2 - r2)/r2^(5/2)]];

rMin = 0.2;

VectorPlot3D[{Bx2[x, y, z], By2[x, y, z], Bz2[x, y, z]},
  {x, -3, 3}, {y, -3, 3}, {z, -3, 3},
  RegionFunction -> Function[{x, y, z}, x^2 + y^2 + z^2 > rMin^2],
  VectorPoints -> 15,
  VectorScale -> {0.03, Scaled[0.6], Automatic},
  VectorColorFunction -> "Rainbow",
  VectorStyle -> Directive[Arrowheads[0.015], Opacity[0.9]],
  PlotLabel -> Style["Campo Magnético Terrestre", 16, Bold, Darker[Blue]],
  AxesLabel -> {Style["x", 14], Style["y", 14], Style["z", 14]},
  AxesStyle -> Thick,
  BoxStyle -> Thick,
  BoxRatios -> {1, 1, 1},
  ViewPoint -> {2.5, 1.8, 1.2},
  ImageSize -> 750]
```
