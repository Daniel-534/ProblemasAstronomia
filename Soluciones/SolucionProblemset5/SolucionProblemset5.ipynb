# Problema 2

## 1

```
Clear["Global`*"];

r1 = m > 170 || m < 160;
r2 = m > 115;
r3 = 80 < m < 200;
Sol = Reduce[r1 && r2 && r3, m]
```

```
NumberLinePlot[Sol, {m, 115, 200},
    PlotStyle -> {Orange, Thick},
    ImageSize->600
 ]
```

## 2

```
Clear["Global`*"];

scale = 1.7*^-27; (* kg / GeV *)

r1 = m > 170 || m < 160;
r2 = m > 115;
r3 = 80 < m < 200;

Solkg = Reduce[
  (r1 && r2 && r3) /. m -> mk/scale,
  mk
]

NumberLinePlot[
  Solkg,
  {mk, 115*scale, 200*scale},
  PlotStyle -> {Orange, Thick},
  ImageSize -> 600
]
```
