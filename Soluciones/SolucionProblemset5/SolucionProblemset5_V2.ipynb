# Problema 1: Campo Magnético de la Tierra

Un campo magnético es más complejo que un campo gravitacional porque posee **polaridad**: todo imán tiene un polo Norte y un polo Sur, y la fuerza experimentada varía según la posición.

## Modelo Dipolar

La intensidad del campo magnético terrestre en las tres direcciones espaciales $(x, y, z)$ se describe mediante el **modelo dipolar**:

$$B_x = \frac{3xz\,M}{r^5}, \qquad B_y = \frac{3yz\,M}{r^5}, \qquad B_z = \frac{(3z^2 - r^2)\,M}{r^5}$$

donde:
- $(x, y, z)$ son coordenadas en **unidades del radio terrestre** $R_\oplus = 6\,371$ km
- $r = \sqrt{x^2 + y^2 + z^2}$ es la distancia al centro de la Tierra (en $R_\oplus$)
- $M = 31\,000$ nT es el **momento dipolar** de la Tierra
- Las componentes $B_i$ se expresan en **nanoTeslas (nT)**

El campo total se obtiene con el **Teorema de Pitágoras en 3D**:
$$|\mathbf{B}| = \sqrt{B_x^2 + B_y^2 + B_z^2}$$

```
Clear["Global`*"];

(* ── Constantes ── *)
M = 31000;          (* Momento dipolar [nT] *)
REarth = 6371;      (* Radio terrestre [km] *)

(* ── Radio al cuadrado (auxiliar) ── *)
r2[x_, y_, z_] := x^2 + y^2 + z^2;

(* ── Componentes del campo magnético dipolar ── *)
Bx[x_, y_, z_] := 3 x z  M / r2[x, y, z]^(5/2);
By[x_, y_, z_] := 3 y z  M / r2[x, y, z]^(5/2);
Bz[x_, y_, z_] := (3 z^2 - r2[x, y, z]) M / r2[x, y, z]^(5/2);

(* ── Vector de campo y magnitud ── *)
Bvec[x_, y_, z_] := {Bx[x, y, z], By[x, y, z], Bz[x, y, z]};
Bmag[x_, y_, z_] := Norm[Bvec[x, y, z]];

(* Verificación simbólica *)
Bvec[x, y, z] // FullSimplify
```

### Inciso (a) — Órbita de satélites de comunicaciones

**Posición:** $(x, y, z) = (7, 0, 0)$, $r = 7$ (en unidades de $R_\oplus$)  
**Distancia física:** $7 \times 6\,371 \approx 44\,600$ km sobre el ecuador terrestre.

Esta posición corresponde aproximadamente a la **órbita geoestacionaria** (GEO) a unos 35 000 km de altitud.

```
(* Inciso (a): satélite en órbita de comunicaciones *)
{xa, ya, za} = {7, 0, 0};

Bxa = Bx[xa, ya, za];
Bya = By[xa, ya, za];
Bza = Bz[xa, ya, za];

Print["Bx = ", N[Bxa, 4], " nT"];
Print["By = ", N[Bya, 4], " nT"];
Print["Bz = ", N[Bza, 4], " nT"];
Print["Distancia fisica: ", 7*REarth, " km ≈ ", N[7*REarth/1000, 3], " Mm"]
```

### Inciso (b) — Cinturones de Van Allen

**Posición:** $(x, y, z) = (0.38, 0.19, 1.73)$, $r = 3$  
**Distancia física:** $3 \times 6\,371 \approx 19\,000$ km.

Los **Cinturones de Van Allen** son regiones de la magnetosfera terrestre donde partículas cargadas quedan atrapadas por el campo magnético. El cinturón exterior se extiende desde aproximadamente 2 hasta 7 radios terrestres.

```
(* Inciso (b): Cinturones de Van Allen *)
{xb, yb, zb} = {0.38, 0.19, 1.73};

Bxb = Bx[xb, yb, zb];
Byb = By[xb, yb, zb];
Bzb = Bz[xb, yb, zb];

Print["Bx = ", N[Bxb, 4], " nT"];
Print["By = ", N[Byb, 4], " nT"];
Print["Bz = ", N[Bzb, 4], " nT"];
Print["Distancia fisica: ", 3*REarth, " km"]
```

### Inciso (c) — Distancia de la Luna

**Posición:** $(x, y, z) = (0, 48, 36)$, $r = 60$  
**Distancia física:** $60 \times 6\,371 \approx 382\,000$ km  
(La distancia media Tierra–Luna es $\approx 384\,400$ km, ✓)

A esta distancia el campo magnético terrestre es muy débil — el campo interplanetario y el del Sol dominan.

```
(* Inciso (c): Distancia de la Luna *)
{xc, yc, zc} = {0, 48, 36};

Bxc = Bx[xc, yc, zc];
Byc = By[xc, yc, zc];
Bzc = Bz[xc, yc, zc];

Print["Bx = ", N[Bxc, 4], " nT"];
Print["By = ", N[Byc, 4], " nT"];
Print["Bz = ", N[Bzc, 4], " nT"];
Print["Distancia fisica: ", 60*REarth, " km"]
```

### Inciso (d) — Magnitud del campo: Teorema de Pitágoras en 3D

La **magnitud absoluta** del campo magnético se obtiene como:

$$|\mathbf{B}| = \sqrt{B_x^2 + B_y^2 + B_z^2}$$

Este es el módulo del vector campo en cada posición.

```
(* Inciso (d): Magnitudes absolutas |B| *)
Bmaga = Bmag[xa, ya, za];
Bmagb = Bmag[xb, yb, zb];
Bmagc = Bmag[xc, yc, zc];

Print["(a) Satelite GEO:       |B| = ", N[Bmaga, 5], " nT"];
Print["(b) Cinturon Van Allen: |B| = ", N[Bmagb, 5], " nT"];
Print["(c) Distancia Luna:     |B| = ", N[Bmagc, 5], " nT"];
```

# Problema 2: El Bosón de Higgs

## Contexto físico

El **bosón de Higgs** es la partícula que, según el Modelo Estándar de la física de partículas, otorga masa a las demás partículas fundamentales a través de su interacción con el **campo de Higgs**.

Antes de su descubrimiento en 2012 en el LHC del CERN, varios experimentos habían impuesto restricciones sobre el rango de su masa posible:

| Experimento | Restricción |
| --- | --- |
| Tevatrón (Fermilab) | $m < 160$ GeV **ó** $m > 170$ GeV |
| LEP (CERN) | $m > 115$ GeV |
| Modelo Estándar (teórico) | $80 < m < 200$ GeV (y además $m < 190$ GeV) |

## Planteamiento

Buscamos el conjunto de valores de masa **consistente con todas las restricciones** simultáneamente.

### Restricciones combinadas:

$$\underbrace{(m < 160 \text{ GeV} \lor m > 170 \text{ GeV})}_{\text{Tevatrón}} \;\cap\; \underbrace{m > 115 \text{ GeV}}_{\text{LEP}} \;\cap\; \underbrace{80 < m < 200 \text{ GeV}}_{\text{MS}}$$

Esto equivale a:
$$115 \text{ GeV} < m < 160 \text{ GeV} \qquad (\text{único intervalo consistente})$$

```
Clear["Global`*"];

(* ── Restricciones sobre la masa del boson de Higgs ── *)

(* Tevatron: m > 170 GeV  ó  m < 160 GeV *)
r1 = m > 170 || m < 160;

(* LEP: m > 115 GeV *)
r2 = m > 115;

(* Modelo Estandar: 80 < m < 200 GeV (con limite adicional m < 190 GeV) *)
r3 = 80 < m < 200;

(* Interseccion de todas las restricciones *)
Sol = Reduce[r1 && r2 && r3, m, Reals]
```

### Inciso (a) — Intersección de restricciones en GeV

La solución anterior nos da el intervalo consistente. Lo graficamos con `NumberLinePlot`.

```
(* Grafico de la recta numerica mostrando cada restriccion *)

(* Colores por restriccion *)
plotTevatron = NumberLinePlot[
  r1, {m, 80, 200},
  PlotStyle -> Directive[Orange, AbsoluteThickness[6]],
  PlotLabel -> Style["Tevatrón: m < 160 ó m > 170 GeV", 11, Orange]
];
plotLEP = NumberLinePlot[
  r2, {m, 80, 200},
  PlotStyle -> Directive[Darker[Blue], AbsoluteThickness[6]],
  PlotLabel -> Style["LEP: m > 115 GeV", 11, Darker[Blue]]
];
plotMS = NumberLinePlot[
  r3, {m, 80, 200},
  PlotStyle -> Directive[Darker[Green], AbsoluteThickness[6]],
  PlotLabel -> Style["Modelo Estándar: 80 < m < 200 GeV", 11, Darker[Green]]
];
plotSol = NumberLinePlot[
  Sol, {m, 80, 200},
  PlotStyle -> Directive[Red, AbsoluteThickness[8]],
  PlotLabel -> Style["Intersección: 115 < m < 160 GeV", 12, Red, Bold]
];

GraphicsColumn[
  {plotTevatron, plotLEP, plotMS, plotSol},
  Spacings -> 10,
  ImageSize -> 600
]
```

### Inciso (b) — Rango de masa en kg

La masa equivalente de $100$ GeV es $1.7 \times 10^{-25}$ kg.  
Por lo tanto, el factor de conversión es:

$$1 \text{ GeV} \equiv \frac{1.7 \times 10^{-25}}{100} = 1.7 \times 10^{-27} \text{ kg/GeV}$$

Aplicando esta conversión al intervalo encontrado $115 < m < 160$ GeV:

$$m_{\min} = 115 \times 1.7 \times 10^{-27} = 1.955 \times 10^{-25} \text{ kg}$$
$$m_{\max} = 160 \times 1.7 \times 10^{-27} = 2.72 \times 10^{-25} \text{ kg}$$

```
(* Factor de conversion: 100 GeV = 1.7e-25 kg *)
scale = 1.7*^-25 / 100;  (* kg/GeV *)
Print["Factor de conversion: ", scale, " kg/GeV"];

(* Rango en kg *)
mMinGeV = 115;  mMaxGeV = 160;
mMinKg  = mMinGeV * scale;
mMaxKg  = mMaxGeV * scale;

Print["Masa minima: ", N[mMinKg, 3], " kg  = ", ScientificForm[N[mMinKg]], " kg"];
Print["Masa maxima: ", N[mMaxKg, 3], " kg  = ", ScientificForm[N[mMaxKg]], " kg"];
Print["\nEl bos\[OAcute]n de Higgs tiene una masa entre: "];
Print[ScientificForm[N[mMinKg]], " kg  y  ", ScientificForm[N[mMaxKg]], " kg"]
```

# Problema 3: Colapso Gravitacional

## Contexto físico

Las **nubes de gas interestelar** son regiones del espacio donde la densidad es suficientemente alta como para que la **gravedad** supere la presión interna, iniciando un colapso que puede llevar a la formación de estrellas.

Este proceso es conocido como **inestabilidad de Jeans**, y el tiempo característico de colapso (también llamado **tiempo de caída libre**) está relacionado con la densidad:

$$T = \frac{2 \times 10^6}{\sqrt{N}}$$

donde:
- $T$ = tiempo de colapso en **años**
- $N$ = densidad de la nube en **átomos/cm³**

**Derivación física:** El tiempo de coída libre proviene de la dinámica gravitacional. Para una esfera de densidad uniforme $\rho$:
$$t_{\rm ff} = \sqrt{\frac{3\pi}{32 G \rho}}$$
Convirtiendo $\rho$ (kg/m³) a $N$ (átomos H/cm³) y el tiempo a años, se obtiene la fórmula del problema.

```
Clear["Global`*"];

(* ── Función de tiempo de colapso ── *)
T[n_] := 2*^6 / Sqrt[n];

(* Verificación con derivacion fisica *)
(* t_ff = Sqrt[3Pi/(32 G rho)], con rho = N * m_H *)
G = 6.674*^-11;      (* m^3 kg^-1 s^-2 *)
mH = 1.6736*^-27;    (* kg, masa del atomo de hidrogeno *)
yrInS = 3.156*^7;    (* segundos por año *)
cm3InM3 = 1*^-6;     (* m^3 por cm^3 *)

(* t_ff en años para N atomos/cm^3 *)
tff[n_] := Sqrt[3 Pi / (32 G (n/cm3InM3 * mH))] / yrInS;

Print["Comparacion de formulas para N=1 atomo/cm^3:"];
Print["  Formula del problema: ", N[T[1], 4], " años"];
Print["  Derivacion fisica:    ", N[tff[1], 4], " años"]
```

### Inciso (a) — Bok Globule

Una **Bok Globule** (en honor al astrónomo Bart Bok) es una nube molecular densa, pequeña y oscura, típicamente con:
- Diámetro: $\sim 1$ año luz
- Masa: $\sim 100 M_\odot$
- Densidad: $N \approx 4\,000$ átomos/cm³

Son las **cunas de formación estelar** más compactas que conocemos.

```
(* Inciso (a): Bok Globule *)
Nbok = 4000;   (* atomos/cm^3 *)
Tbok = T[Nbok];

Print["Bok Globule (N = ", Nbok, " atomos/cm^3):"];
Print["  T = 2e6 / Sqrt[", Nbok, "] = ", N[Tbok, 5], " años"];
Print["  T ≈ ", Round[Tbok, 1], " años"];
Print["  T ≈ ", N[Tbok/1000, 3], " miles de años"];
Print["  (Para referencia: 1 millon de años = 1 Myr)"]
```

### Inciso (b) — Nube primordial de la Vía Láctea

La **Vía Láctea** se formó hace unos 13 000 millones de años a partir de una enorme nube de gas primordial (principalmente hidrógeno y helio). Su densidad era extremadamente baja:
$N \approx 2$ átomos/cm³

Este colapso tan lento es consistente con la formación de grandes estructuras galácticas en tiempos cósmicos.

```
(* Inciso (b): Via Lactea *)
Nmw = 2;  (* atomos/cm^3 *)
Tmw = T[Nmw];

Print["Via Lactea (N = ", Nmw, " atomos/cm^3):"];
Print["  T = 2e6 / Sqrt[", Nmw, "] = ", N[Tmw, 5], " años"];
Print["  T ≈ ", Round[Tmw, 1], " años"];
Print["  T ≈ ", N[Tmw/1*^9, 3], " Gyr (miles de millones de años)"];
Print["  (Edad del universo: ~13.8 Gyr)"]
```

# Resumen y Conclusiones

## Problema 1: Campo Magnético de la Tierra

| Posición | $B_x$ (nT) | $B_y$ (nT) | $B_z$ (nT) | $|\mathbf{B}|$ (nT) |
| --- | --- | --- | --- | --- |
| (a) Satélite GEO, $r=7$ | 0 | 0 | $\approx -20.3$ | $\approx 20.3$ |
| (b) Van Allen, $r=3$ | $\approx 87.2$ | $\approx 43.6$ | $\approx 396$ | $\approx 406$ |
| (c) Luna, $r=60$ | 0 | $\approx 1.43$ | $\approx -0.26$ | $\approx 1.45$ |

El campo cae $\propto r^{-3}$: a la distancia de la Luna es $\sim 10\,000$ veces más débil que en los cinturones de Van Allen.

## Problema 2: Bosón de Higgs

- **Intersección consistente:** $115 \text{ GeV} < m_H < 160 \text{ GeV}$
- **En kg:** $\approx 1.955 \times 10^{-25}$ kg $< m_H < 2.72 \times 10^{-25}$ kg
- El valor experimental real es $m_H = 125.09$ GeV ✓ (dentro del rango)

## Problema 3: Colapso Gravitacional

| Nube | Densidad | Tiempo de colapso |
| --- | --- | --- |
| Bok Globule | $4\,000$ at/cm³ | $\approx 31\,623$ años |
| Vía Láctea | $2$ at/cm³ | $\approx 1.41 \times 10^6$ años (1.41 Myr) |

La ley $T \propto N^{-1/2}$ refleja directamente el tiempo de caída libre de Jeans: nubes más densas colapsan mucho más rápido.
