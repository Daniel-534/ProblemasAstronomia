## Problema 1 — Agujeros Negros Estelares más Cercanos

La Vía Láctea es un disco plano de estrellas, por lo que podemos usar coordenadas cartesianas para
localizar los agujeros negros más cercanos al Sol.  
La tabla proporcionada contiene:
- **Nombre** del sistema binario de rayos X
- **Masa** del agujero negro (en masas solares $M_\odot$)
- **Coordenadas** $(x, y)$ en años luz (al), con el Sol en el origen

**Tareas:**
1. Trazar la cuadrícula cartesiana con la posición de cada agujero negro.
2. Calcular la distancia de cada agujero negro al Sol.
3. Calcular la distancia **media**, **mediana** y **moda**.

```
(* ──────────────────────────────────────────────────────────────────────── *)
(*  DATOS — Tabla de agujeros negros estelares más cercanos al Sol          *)
(* ──────────────────────────────────────────────────────────────────────── *)

nombres = {"Cygnus X-1", "SS-433", "Nova Monocerotis 1975",
            "Nova Persei 1992", "IL Lupi", "Nova Vulpeculi 1988", "V404 Cygni"};

letras  = {"A", "B", "C", "D", "E", "F", "G"};
masas   = {16, 11, 11, 5, 9, 8, 12};         (* M_sol *)
coordX  = {6600, 8000, -1400, 5600, 6500, 2200, 6900}; (* años luz *)
coordY  = {-2400, -14000, 2400, 3300, -11000, -6100, -4000};
coords  = Transpose[{coordX, coordY}];
n       = Length[nombres];

Print["Datos cargados: ", n, " agujeros negros."]
```

### Parte 1 — Cuadrícula cartesiana de los agujeros negros

Se construye un mapa tipo «espacio profundo» con el Sol en el origen $(0, 0)$.  
El **tamaño** de cada punto es proporcional a la masa del agujero negro y el **color** va
de azul (masa baja) a rojo-naranja (masa alta).

```
(* ── Escala de color por masa (azul → naranja-rojo) ── *)
masaMin = Min[masas];  masaMax = Max[masas];
bhColor[i_] := Blend[{accentBl, accentOr}, (masas[[i]] - masaMin)/(masaMax - masaMin)];

(* ── Radio de cada disco proporcional a la masa ── *)
disco[i_] := 180 + 55 masas[[i]];

(* ── Offsets para que las etiquetas no se superpongan ── *)
offsets = {{420, 420}, {420, 420}, {-420, 420}, {420, 420},
            {420, -480}, {-420, -480}, {420, 420}};

(* ── Construcción del mapa ── *)
mapa = Graphics[
  {
    (* Fondo espacial *)
    {bgDark, Rectangle[{-16600, -16600}, {16600, 16600}]},

    (* Líneas de cuadrícula *)
    {gridCol, Dashed, AbsoluteThickness[0.5],
     Table[Line[{{x, -16000}, {x, 16000}}], {x, -15000, 15000, 3000}],
     Table[Line[{{-16000, y}, {16000, y}}], {y, -15000, 15000, 3000}]},

    (* Ejes principales *)
    {axisCol, AbsoluteThickness[1.2],
     Line[{{-16000, 0}, {16000, 0}}],
     Line[{{0, -16000}, {0, 16000}}]},

    (* Marcas numéricas en los ejes *)
    {axisCol,
     Table[
       {Line[{{x, -180}, {x, 180}}],
        Text[Style[ToString[x/1000] <> "k", 8, axisCol], {x, -600}]},
       {x, -15000, 15000, 3000}],
     Table[
       {Line[{{-180, y}, {180, y}}],
        Text[Style[ToString[y/1000] <> "k", 8, axisCol], {-900, y}]},
       {y, -15000, 15000, 3000}]},

    (* Sol *)
    {sunYellow, Glow[sunYellow], Disk[{0, 0}, 320]},
    {white, Text[Style["\[FivePointedStar] Sol", 11, Bold, white], {0, -780}]},

    (* Agujeros negros *)
    Table[
      {
        (* Halo brillante *)
        {Opacity[0.25], bhColor[i], Disk[coords[[i]], disco[i] * 1.6]},
        (* Disco central *)
        {bhColor[i], Disk[coords[[i]], disco[i]]},
        (* Etiqueta *)
        Text[
          Style[letras[[i]] <> ") " <> nombres[[i]], 9, white, Bold],
          coords[[i]] + offsets[[i]]
        ]
      },
      {i, 1, n}
    ],

    (* Leyenda de masa *)
    {white, Text[Style["\[FilledCircle] masa baja", 9, accentBl], {-12800, 14800}]},
    {white, Text[Style["\[FilledCircle] masa alta", 9, accentOr], {-12800, 13900}]}
  },
  PlotRange   -> {{-16600, 16600}, {-16600, 16600}},
  Background  -> bgDark,
  ImageSize   -> {700, 700},
  PlotLabel   -> Style["Agujeros Negros Estelares m\[AAcute]s Cercanos al Sol",
                       white, 14, Bold],
  Frame       -> True,
  FrameStyle  -> axisCol,
  FrameLabel  -> {
    {Style["y  (a\[NTilde]os luz)", axisCol, 12], None},
    {Style["x  (a\[NTilde]os luz)", axisCol, 12], None}
  }
]

mapa
```

### Parte 2 — Distancia de cada agujero negro al Sol

La distancia euclidiana desde el origen se calcula como:

$$d = \sqrt{x^2 + y^2}$$

```
(* ── Cálculo de distancias ── *)
distancias = N[Sqrt[coordX[[#]]^2 + coordY[[#]]^2]] & /@ Range[n];

(* ── Tabla de resultados ── *)
filas = Table[
  {letras[[i]], nombres[[i]], masas[[i]], coordX[[i]], coordY[[i]],
   Row[{NumberForm[distancias[[i]], {7, 1}], " al"}]},
  {i, 1, n}
];

header = {Style["Letra", Bold], Style["Nombre", Bold],
           Style["Masa (M\[SunIcon])", Bold],
           Style["x (al)", Bold], Style["y (al)", Bold],
           Style["d (al)", Bold]};

Grid[
  Prepend[filas, header],
  Frame     -> All,
  Alignment -> {Left, Center},
  Background -> {None, {LightGray, {White}}},
  Spacings  -> {2, 1.2}
]
```

### Parte 3 — Estadísticas: media, mediana y moda

Con las siete distancias calculadas se obtienen las medidas de tendencia central.

```
(* ── Estadísticas ── *)
media   = N@Mean[distancias];
mediana = N@Median[distancias];

(* La moda requiere valores repetidos; en este caso ninguna distancia se repite *)
frecuencias = Tally[Round[distancias, 1]];
modaVal     = Select[frecuencias, #[[2]] == Max[frecuencias[[All, 2]]] &];

Print["Distancias (años luz):"];
Print[NumberForm[#, {8, 1}] & /@ distancias];
Print[""];
Print["Media   = ", NumberForm[media,   {8, 1}], " al"];
Print["Mediana = ", NumberForm[mediana, {8, 1}], " al"];
Print["Moda    = ",
  If[Length[modaVal] == Length[distancias],
    "No existe moda \[LongDash] todas las distancias son distintas.",
    Row[{NumberForm[modaVal[[1, 1]], {8, 1}], " al"}]
  ]
];
```

```
(* ── Diagrama de puntos de las distancias (con media y mediana) ── *)
coloresPuntos = Table[bhColor[i], {i, 1, n}];

Show[
  Graphics[
    {
      {bgDark, Rectangle[{500, -0.2}, {17000, 1.2}]},

      (* Línea de media *)
      {Dashed, accentOr, AbsoluteThickness[1.5],
       Line[{{media, -0.15}, {media, 1.15}}]},
      {accentOr, Text[Style["Media", 9, accentOr], {media, 1.22}]},

      (* Línea de mediana *)
      {Dashed, accentBl, AbsoluteThickness[1.5],
       Line[{{mediana, -0.15}, {mediana, 1.15}}]},
      {accentBl, Text[Style["Mediana", 9, accentBl], {mediana, -0.23}]},

      (* Puntos *)
      Table[
        {
          {Opacity[0.3], coloresPuntos[[i]],
           Disk[{distancias[[i]], 0.5}, {200, 0.06}]},
          {coloresPuntos[[i]],
           Disk[{distancias[[i]], 0.5}, {120, 0.04}]},
          {white,
           Text[Style[letras[[i]], 8, white, Bold],
                {distancias[[i]], 0.5 + 0.16}]}
        },
        {i, 1, n}
      ]
    },
    PlotRange  -> {{500, 17000}, {-0.4, 1.5}},
    Background -> bgDark,
    ImageSize  -> {700, 240},
    PlotLabel  -> Style["Distribución de distancias al Sol", white, 13, Bold],
    Frame      -> True,
    FrameStyle -> axisCol,
    FrameLabel -> {{None, None}, {Style["Distancia (años luz)", axisCol, 11], None}},
    FrameTicks -> {{{}, {}},
                   {Range[0, 17000, 2000], None}}
  ]
]
```

## Problema 2 — Ley Masa–Radio de los Agujeros Negros

Los agujeros negros son extraordinariamente densos: enormes cantidades de materia caben en
volúmenes muy pequeños.  La tabla siguiente relaciona la **masa** (en masas de la Tierra $M_\oplus$)
con el **radio de Schwarzschild** predicho (en cm).

| Masa ($M_\oplus$) | Radio (cm) |
|:-----------------:|:----------:|
| 2.0  | 16.8 |
| 3.2  | 26.9 |
| 5.0  | 42.0 |
| 7.5  | 63.0 |
| 8.7  | 73.1 |
| 11.0 | 96.6 |

**Tareas:**
1. Graficar los datos.
2. Estimar la pendiente desde la gráfica.
3. Calcular la pendiente desde la tabla.
4. Escribir la ecuación lineal $r(m) = s\,m + r_0$.
5. Predecir el radio de un agujero negro con la masa de Júpiter ($318\,M_\oplus$).

```
(* ── Datos ── *)
datosMR = {{2.0, 16.8}, {3.2, 26.9}, {5.0, 42.0},
            {7.5, 63.0}, {8.7, 73.1}, {11.0, 96.6}};
masasMR  = datosMR[[All, 1]];   (* M_tierra *)
radios   = datosMR[[All, 2]];   (* cm       *)
Print["Datos cargados: ", Length[datosMR], " puntos."]
```

### Parte 1 — Gráfico de los datos

```
(* ── Ajuste lineal (para superponer la recta) ── *)
ajuste = LinearModelFit[datosMR, m, m];
pendiente  = ajuste["BestFitParameters"][[2]];
intercepto = ajuste["BestFitParameters"][[1]];

(* ── Gráfico completo ── *)
graficoDatos = Show[
  (* Recta de ajuste *)
  Plot[
    ajuste[m], {m, 0, 12},
    PlotStyle -> {Thick, accentOr},
    PlotLegends -> {Style["Ajuste lineal", 11]}
  ],
  (* Datos crudos *)
  ListPlot[
    datosMR,
    PlotStyle    -> {PointSize[0.018], accentBl},
    PlotLegends  -> {Style["Datos observados", 11]}
  ],
  (* Líneas verticales de residuo *)
  Graphics[
    Table[
      {GrayLevel[0.6], Dashed, AbsoluteThickness[0.8],
       Line[{{masasMR[[i]], radios[[i]]}, {masasMR[[i]], ajuste[masasMR[[i]]]}}]},
      {i, 1, Length[datosMR]}
    ]
  ],
  (* Ecuación sobre la gráfica *)
  Graphics[
    {accentOr,
     Text[
       Style[
         Row[{"r(m) = ",
              NumberForm[pendiente, {5, 2}], " m + ",
              NumberForm[intercepto, {5, 2}]}],
         12, Bold, accentOr
       ],
       {2.2, 88}
     ]
    }
  ],
  Background -> GrayLevel[0.97],
  Frame      -> True,
  FrameStyle -> GrayLevel[0.3],
  FrameLabel -> {
    {Style["Radio de Schwarzschild  r  (cm)", 12], None},
    {Style["Masa  m  (M\[CirclePlus])", 12], None}
  },
  PlotLabel -> Style["Ley Masa\[Dash]Radio de los Agujeros Negros", 14, Bold],
  ImageSize -> {650, 450},
  GridLines -> Automatic,
  GridLinesStyle -> {Dashed, GrayLevel[0.8]}
]

graficoDatos
```

### Partes 2 y 3 — Cálculo de la pendiente

**Desde la gráfica:** se elige un intervalo amplio y se calcula $s = \Delta r / \Delta m$.

**Desde la tabla:** la forma más rigurosa es ajustar una recta por mínimos cuadrados.

```
(* ── Pendiente visual (primer y último punto) ── *)
sVisual = (radios[[-1]] - radios[[1]]) / (masasMR[[-1]] - masasMR[[1]]);
Print["Pendiente visual (\[CapitalDelta]r/\[CapitalDelta]m, extremos de la tabla): ",
      NumberForm[sVisual, {6, 3}], " cm / M\[CirclePlus]"];

(* ── Pendiente por mínimos cuadrados ── *)
Print[""];
Print["Parámetros del ajuste lineal (mínimos cuadrados):"];
Print["  Intercepto  r\[Subscript]0 = ", NumberForm[intercepto, {6, 3}], " cm"];
Print["  Pendiente   s  = ",  NumberForm[pendiente,  {6, 3}], " cm / M\[CirclePlus]"];
Print[""];
Print["Unidades de la pendiente: centímetros por masa terrestre  [cm / M\[CirclePlus]]"];
Print[""];
Print["Bondad del ajuste  R\[Squared] = ", NumberForm[ajuste["RSquared"], {5, 5}]];
```

### Parte 4 — Ecuación lineal $r(m) = s\,m + r_0$

```
(* ── Ecuación formal ── *)
rFun[m_] := pendiente * m + intercepto;

Print["Ley Masa-Radio del agujero negro:"];
Print[""];
Print["  r(m)  =  ",
      NumberForm[pendiente, {5, 3}],
      "  m  +  ",
      NumberForm[intercepto, {5, 3}]];
Print[""];
Print["donde  m  es la masa en unidades de M\[CirclePlus]  y  r  el radio en cm."];

(* ── Verificación ── *)
Print[""];
Print["Verificación (predichos vs observados):"];
Grid[
  Prepend[
    Table[
      {masasMR[[i]], radios[[i]],
       NumberForm[rFun[masasMR[[i]]], {6, 2}],
       NumberForm[radios[[i]] - rFun[masasMR[[i]]], {6, 2}]},
      {i, 1, Length[datosMR]}
    ],
    {Style["m (M\[CirclePlus])", Bold], Style["r obs (cm)", Bold],
     Style["r pred (cm)", Bold], Style["residuo (cm)", Bold]}
  ],
  Frame -> All, Spacings -> {2, 1},
  Background -> {None, {LightGray, {White}}}
]
```

### Parte 5 — Predicción para Júpiter

La masa de Júpiter es $318\,M_\oplus$, bien por fuera del rango de ajuste, por lo que
se aplica la ecuación obtenida como extrapolación.

```
mJupiter = 318.0;   (* M_tierra *)

rJupiter = rFun[mJupiter];

Print["Masa de Júpiter: ", mJupiter, " M\[CirclePlus]"];
Print["Radio predicho del agujero negro: ",
      NumberForm[rJupiter, {8, 2}], " cm"];
Print["                                  ≈ ",
      NumberForm[rJupiter/100, {8, 2}], " m"];
Print[""];
Print["Para referencia, un balón de fútbol mide ~22 cm de diámetro."];
Print["Radio de Schwarzschild de Júpiter ≈ ",
      NumberForm[rJupiter, {7, 0}], " cm  ≈  ",
      NumberForm[rJupiter / 22, {5, 1}],
      " veces el radio de un balón de fútbol."];
```

```
(* ── Gráfico con extrapolación hasta Júpiter ── *)
Show[
  Plot[
    rFun[m], {m, 0, 330},
    PlotStyle  -> {Thick, accentOr},
    PlotLegends -> {Style["Extrapolación", 11]}
  ],
  ListPlot[
    datosMR,
    PlotStyle   -> {PointSize[0.016], accentBl},
    PlotLegends -> {Style["Datos tabla", 11]}
  ],
  ListPlot[
    {{mJupiter, rJupiter}},
    PlotStyle   -> {PointSize[0.020], Red},
    PlotLegends -> {Style["Júpiter (predicho)", 11]}
  ],
  Graphics[
    {
      {Red, Dashed, AbsoluteThickness[0.9],
       Line[{{mJupiter, 0}, {mJupiter, rJupiter}}],
       Line[{{0, rJupiter}, {mJupiter, rJupiter}}]},
      {Red, Text[
        Style[Row[{"Júpiter: ", NumberForm[rJupiter, {7, 1}], " cm"}], 10, Bold, Red],
        {mJupiter - 30, rJupiter + 40}
      ]}
    }
  ],
  Background -> GrayLevel[0.97],
  Frame      -> True,
  FrameStyle -> GrayLevel[0.3],
  FrameLabel -> {
    {Style["Radio  r  (cm)", 12], None},
    {Style["Masa  m  (M\[CirclePlus])", 12], None}
  },
  PlotLabel  -> Style["Extrapolación hasta la masa de Júpiter", 14, Bold],
  ImageSize  -> {650, 420},
  GridLines  -> Automatic,
  GridLinesStyle -> {Dashed, GrayLevel[0.8]}
]
```

## Problema 3 — Temperaturas del Disco de Acreción

Cuando materia cae hacia un agujero negro, su energía potencial gravitatoria se convierte en
energía cinética y luego en calor por fricción.  Igualando la energía potencial gravitatoria
con la energía térmica de una partícula obtenemos:

$$E_P = E_T \quad \Longrightarrow \quad \frac{GMm}{R} = \frac{3}{2}kT$$

$$\boxed{\; T = \frac{2\,G\,M\,m}{3\,k\,R} \;}$$

donde:
- $G = 6.67 \times 10^{-11}\,\text{N m}^2/\text{kg}^2$
- $M = 1.9 \times 10^{30}\,\text{kg}$ (masa solar)
- $m = 1.6 \times 10^{-27}\,\text{kg}$ (masa del protón / hidrógeno)
- $k = 1.38 \times 10^{-23}\,\text{J/K}$ (constante de Boltzmann)
- $R$ = distancia al agujero negro [m]

**Tareas:**
1. Temperatura a la distancia de la órbita terrestre ($R = 1.47 \times 10^{11}\,\text{m}$).
2. Temperatura a la distancia de Neptuno ($R = 4.4 \times 10^{12}\,\text{m}$).
3. ¿Qué tipo de radiación domina en cada zona del disco?

```
(* ── Constantes y parámetros ── *)
GG  = 6.67 * 10^-11;   (* N m^2 / kg^2 *)
MM  = 1.9  * 10^30;    (* kg, masa solar *)
mp  = 1.6  * 10^-27;   (* kg, masa protón (hidrógeno) *)
kB  = 1.38 * 10^-23;   (* J / K, constante de Boltzmann *)

(* ── Función temperatura *)
Temp[R_] := (2 GG MM mp) / (3 kB R);

Print["Función T(R) definida correctamente."];
Print["T[R] = 2 G M m / (3 k R)"];
```

### Parte 1 — Temperatura a la distancia de la órbita de la Tierra

$R_\oplus = 1.47 \times 10^{11}\,\text{m}$

```
RTierra = 1.47 * 10^11;   (* m *)

tTierra = Temp[RTierra];

Print["T(R_Tierra) = ", NumberForm[tTierra, {8, 0}], " K"];
Print["           = ", NumberForm[tTierra/1000, {7, 1}], " × 10^3 K"];
Print[""];
Print["Clasificación:"];
Which[
  tTierra > 100000,
    Print["  ► T > 100 000 K  →  RAYOS X"],
  tTierra > 2000,
    Print["  ► 2 000 K < T < 100 000 K  →  LUZ VISIBLE"],
  True,
    Print["  ► T < 2 000 K  →  INFRARROJO"]
];
```

### Parte 2 — Temperatura a la distancia de la órbita de Neptuno

$R_\text{Nep} = 4.4 \times 10^{12}\,\text{m}$

```
RNeptuno = 4.4 * 10^12;   (* m *)

tNeptuno = Temp[RNeptuno];

Print["T(R_Neptuno) = ", NumberForm[tNeptuno, {8, 2}], " K"];
Print[""];
Print["Clasificación:"];
Which[
  tNeptuno > 100000,
    Print["  ► T > 100 000 K  →  RAYOS X"],
  tNeptuno > 2000,
    Print["  ► 2 000 K < T < 100 000 K  →  LUZ VISIBLE"],
  True,
    Print["  ► T < 2 000 K  →  INFRARROJO"]
];
```

### Parte 3 — Tipo de radiación dominante y perfil térmico del disco

| Región | Distancia | Temperatura | Emisión dominante |
|--------|-----------|-------------|-------------------|
| Órbita de Mercurio ($5.8\times10^{10}$ m) | 0.4 UA | ~167 000 K | **Rayos X** |
| Órbita de la Tierra | 1 UA | ~66 600 K | **Rayos X** |
| Órbita de Marte ($2.3\times10^{11}$ m) | 1.5 UA | ~42 500 K | **Rayos X / UV** |
| Disco intermedio | 5–10 UA | 3 000–13 000 K | **Luz visible** |
| Órbita de Neptuno | 30 UA | ~2 226 K | **Visible / Infrarrojo** |

**Conclusión:** el disco de acreción de un agujero negro de masa solar emitirá principalmente
**rayos X** en sus regiones internas (donde el gas es más caliente y denso) y
**luz visible** en las regiones externas.  Muy lejos del agujero negro (tipo órbita de Neptuno),
el gas alcanzará temperaturas inferiores a $2\,500\,\text{K}$, emitiendo en el **infrarrojo cercano**.
Esto es consistente con las observaciones de fuentes binarias de rayos X como Cygnus X-1.

```
(* ── Perfil de temperatura del disco (escala log-log) ── *)

rMin = 5.8 * 10^10;    (* ~Mercurio *)
rMax = 5.0 * 10^12;    (* más allá de Neptuno *)

(* Zonas de emisión *)
zonas = {
  {2000,   100000, RGBColor[0.20, 0.60, 1.0],  "Luz visible"},
  {100000, 10^8,   RGBColor[0.90, 0.20, 0.20], "Rayos X"  },
  {500,    2000,   RGBColor[0.80, 0.50, 0.10], "Infrarrojo"}
};

perfilT = LogLogPlot[
  Temp[R], {R, rMin, rMax},
  PlotStyle  -> {AbsoluteThickness[2.5], accentOr},
  Filling    -> Bottom,
  FillingStyle -> Directive[Opacity[0.15], accentOr],
  Frame      -> True,
  FrameStyle -> GrayLevel[0.3],
  Background -> GrayLevel[0.97],
  FrameLabel -> {
    {Style["Temperatura  T  (K)", 12], None},
    {Style["Distancia al agujero negro  R  (m)", 12], None}
  },
  PlotLabel -> Style["Perfil de temperatura del disco de acreción (M = M\[SunIcon])",
                     14, Bold],
  GridLines  -> Automatic,
  GridLinesStyle -> {Dashed, GrayLevel[0.8]},
  ImageSize  -> {680, 420},
  Epilog -> {
    (* Línea T = 100 000 K *)
    {Dashed, Red, AbsoluteThickness[1.2],
     InfiniteLine[{rMin, Log[10, 100000]}, {1, 0}]},
    (* Línea T = 2 000 K *)
    {Dashed, accentBl, AbsoluteThickness[1.2],
     InfiniteLine[{rMin, Log[10, 2000]}, {1, 0}]}
  }
];

(* ── Marcadores de planetas ── *)
planetas = {
  {1.47 * 10^11, "Tierra", accentBl},
  {4.4  * 10^12, "Neptuno", accentBl}
};

marcadores = Graphics[{
  Table[
    {
      {planetas[[i, 3]], PointSize[0.015],
       Point[{Log[10, planetas[[i, 1]]], Log[10, Temp[planetas[[i, 1]]]]}]},
      {planetas[[i, 3]],
       Text[
         Style[planetas[[i, 2]], 10, Bold, planetas[[i, 3]]],
         {Log[10, planetas[[i, 1]]] + 0.12,
          Log[10, Temp[planetas[[i, 1]]]] + 0.18}
       ]}
    },
    {i, 1, Length[planetas]}
  ]
}];

Show[perfilT, marcadores]
```
