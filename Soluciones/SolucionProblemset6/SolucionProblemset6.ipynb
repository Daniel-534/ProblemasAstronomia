# Problema 6

```
Clear["Global`*"]
WolframColor = RGBColor[0.941, 0.502, 0.0];

ruta = FileNameJoin[{Directory[], "BlackHoleCoordinates.dat"}];
datosRaw = Import[ruta, "Table"];

nombres = datosRaw[[All, 1]];
puntos = datosRaw[[All, {2, 3}]];
```

## 1

```
grafica = ListPlot[
Table[
      Callout[puntos[[i]], Style[nombres[[i]], 14]], 
   {i, Length[nombres]}
  ],
  PlotStyle -> {WolframColor, PointSize[0.02]},
  PlotTheme -> "Scientific",
  Frame -> True,
  GridLines -> Automatic,
  FrameLabel ->{
      Style["X [ly]",18,Bold],
      Style["y [ly]",18,Bold]
              },
  FrameTicksStyle -> Directive[14],
  PlotLabel -> Style["Mapa de Agujeros Negros Cercanos al Sol", 18, Bold],
  Epilog -> {Black, PointSize[0.03], Point[{0, 0}], Text[Style["SOL", Bold, Black], {0, 0}, {0, -2}]},
  ImageSize -> 700
]
```

## 2

```
distancias = Norm /@ puntos; 

tablaFinal = Table[
  {
   nombres[[i]], 
   puntos[[i, 1]], 
   puntos[[i, 2]], 
   Round[distancias[[i]], 0.01]
  }, 
  {i, Length[nombres]}
];

Print[TableForm[tablaFinal, 
  TableHeadings -> {None, {"Nombre", "x (ly)", "y (ly)", "d (ly)"}}]];
```

## 3

```
media = Mean[distancias] // N
mediana = Median[distancias] // N 
moda = Commonest[distancias] (* No hay moda *)
```
