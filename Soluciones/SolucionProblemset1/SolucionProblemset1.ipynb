# Problema 2

```python
import math

c = 2.9979e10    # Velocidad de la Luz (cm/s)
h = 6.6262e-27   # Constante de Planck (erg * s)
m = 9.1095e-28   # Masa del Electrón (g)
e = 4.80325e-10  # Carga del Electrón (esu)
G = 6.6732e-8    # Constante de Gravitación (dyn * cm^2 / g^2)
M = 1.6726e-24   # Masa del Protón (g)

def solve_problem_2():
    # 1. Constante de entropía de agujero negro
    c1 = (c**3) / (2 * h * G)

    # 2. Constante de radiación gravitacional
    c2 = (32 * (G**5)) / (5 * (c**10))

    # 3. Constante de Thomas-Fermi [2]
    c3 = (324 / 175) * (4 / (9 * math.pi))**(2/3)

    # 4. Sección transversal de dispersión de Thompson
    c4 = (8 * math.pi / 3) * ( (e**2) / (m * (c**2)) )**2
    
    # 5. Límite de Stark
    c5 = (1 / (M**5)) * ( (4 * (math.pi**2) * (e**2) * m) / (h**2) )**2

    # 6. Constante de radiación de Bremstrahlung
    c6 = (32 * (math.pi**2) * (e**6)) / (3 * math.sqrt(2 * math.pi * (m**3)) * c)

    # 7. Constante de Fotoionización
    c7 = (32 * (math.pi**2) * (e**6) * (2 * (math.pi**2) * (e**4) * m)) / ((3**1.5) * (h**3))

    print(f"1. Entropía de agujero negro: {c1:.2e}")
    print(f"2. Radiación gravitacional: {c2:.2e}")
    print(f"3. Thomas-Fermi: {c3:.2e}")
    print(f"4. Dispersión de Thompson: {c4:.2e}")
    print(f"5. Límite de Stark: {c5:.2e}")
    print(f"6. Radiación de Bremstrahlung: {c6:.2e}")
    print(f"7. Fotoionización: {c7:.2e}")

solve_problem_2()
```

# Problema 3

## 3.1

$$F_g = F_c \Rightarrow M = \frac{R v^2}{G}$$

## 3.2

$$v = \frac{2\pi R}{T}\Rightarrow M = \left(\frac{2\pi}{T}\right)^2\cdot\frac{R^3}{G}$$

## 3.3

```python
import math

def Solve33():
    G = 6.67e-11 # SI
    R = 1737 * 1000 # m
    T = 2 * 60 * 60 # s
    
    M = (2*math.pi/T)**2 * R**3/G
    
    return M
print(f"M = {Solve33():.2e} kg")
```
